```python
import glob, os
import pandas as pd

path = glob.glob("/workspaces/**/HistoricalData_*.csv", recursive=True)[0]
print("Found:", path)

df = pd.read_csv(path)
df = df.iloc[::-1]
df["Close/Last"] = df["Close/Last"].str.replace("$", "", regex=False).astype(float)

out = os.path.join(os.path.dirname(path), "hw4")
os.makedirs(out, exist_ok=True)
df["Close/Last"].to_csv(os.path.join(out, "TSLA.txt"), index=False, header=False)
print("Saved to:", out)


import glob

# Find TSLA.txt in the hw4 folder
file_path = glob.glob("/workspaces/**/hw4/TSLA.txt", recursive=True)[0]

# Read all prices into a list (one price per line)
with open(file_path) as file:
    prices = [round(float(line), 2) for line in file]

buy = None          # price we bought at (None means we don't own the stock)
first_buy = None    # the very first price we bought at
total_profit = 0    # running total of all trade profits

print("TSLA Mean Reversion Strategy Output")

# Start at index 5 so there are 5 previous days to average
for i in range(5, len(prices)):
    current_price = prices[i]
    average = sum(prices[i - 5:i]) / 5   # average of the previous 5 days

    if buy is None and current_price < average * 0.98:
        # Buy low
        buy = current_price
        if first_buy is None:
            first_buy = buy
        print("buying at:      ", buy)
    elif buy is not None and current_price > average * 1.02:
        # Sell high
        trade_profit = round(current_price - buy, 2)
        total_profit += trade_profit
        print("selling at:     ", current_price)
        print("trade profit:   ", trade_profit)
        buy = None
    else:
        # Do nothing this iteration
        pass

final_profit_percentage = (total_profit / first_buy) * 100

print("-----------------------")
print("Total profit:   ", round(total_profit, 2))
print("First buy:      ", first_buy)
print("% return:       ", str(round(final_profit_percentage, 2)) + "%")


```
